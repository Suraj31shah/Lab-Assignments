### I) Marks Analysis

In the **'Marks.csv'** file, you can find the scores obtained by 200 students in 4 subjects of a standardised test. The different columns - `Score A`, `Score B`, `Score C` and `Score D` indicate the score obtained by a particular student in the respective subjects A, B, C and D.



Load the dataset to your notebook and answer the following questions

```python
#Load the necessary Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Load the dataset
df1 = pd.read_csv("Marks.csv")
df1.head()
```

**Q1)** Load the dataset and plot a histogram for the `Score A` column by keeping the `number of bins to 6`. Which bin range among the following has the highest frequency?

(`Note` - *The bin ranges mentioned in the options are approximate values for the bin ranges that you'll actually get when you plot the histogram*)

    a)0-50
    b)50-100
    c)150-200
    d)200-250

```python
#Your code here
plt.hist(df1['Score A'], bins=10)
plt.xlabel("Score A")
plt.ylabel("Frequency")
plt.title("Histogram of Score A")
plt.show()
```

**Q2)** Plot a box plot for the column `Score C` and choose the correct option.

    A - The 25th percentile lies between 20 and 40
    B - The 75th percentile lies between 40 and 60
    C - The 25th percentile lies between 0 and 20
    D - Both B and C

```python
#Your code here
plt.boxplot([df1['Score A'], df1['Score B'], df1["Score C"], df1["Score D"]], labels=["Score A", "score B", "Score C", "Score D"])
plt.title("Boxplot of Scores")
plt.ylabel("Marks")
plt.show()
```

### II) Superstore Data

In the `superstore.csv` file, you have the details of orders purchased in an American online retail store. Load the dataset, observe and analyse the different columns and answer the following questions.

```python
#Load the dataset
df2 =pd.read_csv("superstore.csv")
df2.head()
```

**Q4)** Plot a pie-chart to find the Ship Mode through which most of the orders are being delivered.

     a)Standard Class
     b)First Class
     c)Second Class
     d)Same Day
     

```python
#Your code here
ship_mode_counts=df2["Ship Mode"].value_counts()
plt.pie(ship_mode_counts.values, labels=ship_mode_counts.index, autopct="%1.1f%%", startangle=140)
plt.title("Distribution of orders by ship mode")
plt.show()
```

**Q5)** Plot a bar chart comparing the average `Discount` across all the `Regions` and report back the `Region` getting the highest average discount

**Note** - You need to clean the `Discount` column first

    a)Central
    b)South
    c)West
    d)East

```python
#Your code here
df2["Discount"]=df2["Discount"].astype(str).str.replace("%", "")
df2["Discount"]=df2["Discount"].astype(float)
avg_discount=df2.groupby("Region")["Discount"].mean()
plt.bar(avg_discount.index, avg_discount.values)
plt.xlabel("Region")
plt.ylabel("Average-Discount")
plt.title("Average Discount by Region")
plt.show()
```
