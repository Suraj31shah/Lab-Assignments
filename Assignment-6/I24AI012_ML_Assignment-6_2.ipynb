```python
import pandas as pd
import numpy as np
```

```python
data={
    "Category": ["Parent", "Male", "Female", "Science", "Arts"],
    "Interested": [20, 10, 10, 10, 10],
    "Not_Interested": [30, 20, 10, 30, 0],
    "Total": [50, 30, 20, 40, 10]
}

df=pd.DataFrame(data)
df
```

```python
def gini_from_row(row):
    p1=row["Interested"]/row["Total"]
    p2=row["Not_Interested"]/row["Total"]
    return 1-(p1**2+p2**2)
```

```python
df["Gini"]=df.apply(gini_from_row, axis=1)
df
```

```python
parent_gini=df[df["Category"]=="Parent"]["Gini"].values[0]
print("Gini (Parent):", round(parent_gini, 3))
```

```python
male=df[df["Category"]=="Male"].iloc[0]
female=df[df["Category"]=="Female"].iloc[0]
total = df[df["Category"] == "Parent"]["Total"].values[0]

weighted_gender=(
    (male["Total"]/total)*male["Gini"]+(female["Total"]/total)*female["Gini"]
)

print("Weighted Gini (Gender Split):", round(weighted_gender, 3))
```

```python
science=df[df["Category"]=="Science"].iloc[0]
arts=df[df["Category"]=="Arts"].iloc[0]

weighted_stream=(
    (science["Total"]/total)*science["Gini"]+(arts["Total"]/total)*arts["Gini"]
)

print("Weighted Gini (Stream Split):", round(weighted_stream, 3))
```

```python
print("Parent Gini:", round(parent_gini, 3))
print("Weighted Gender Gini:", round(weighted_gender, 3))
print("Weighted Stream Gini:", round(weighted_stream, 3))

if weighted_stream<weighted_gender:
    print("Better Split: Stream")
else:
    print("Better Split: Gender")
```
